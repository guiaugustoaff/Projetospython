```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv("Games Awards Goty Nominees and Winners 2014-2024.csv", encoding='latin-1')
winners = df[pd.to_numeric(df['Wins']) == 1]

winners.head()
```

```python

year_candidates = ['Year of the Event']
meta_candidates = ['Meta-Score']
user_candidates = ['User-Score']

year_col = next((c for c in year_candidates if c in df.columns), None)
meta_col = next((c for c in meta_candidates if c in df.columns), None)
user_col = next((c for c in user_candidates if c in df.columns), None)

if not year_col or not meta_col or not user_col:
    print('Não encontrei as colunas Year/Meta/User. Colunas disponíveis:')
    print(df.columns.tolist())
else:
    # Mudar para True se quiser analisar apenas os vencedores
    only_winners = False
    data = winners if only_winners else df

    df_year = data[[year_col, meta_col, user_col]].copy()
    df_year[year_col] = pd.to_numeric(df_year[year_col], errors='coerce')
    df_year[meta_col] = pd.to_numeric(df_year[meta_col], errors='coerce')
    df_year[user_col] = pd.to_numeric(df_year[user_col], errors='coerce')
    df_year = df_year.dropna()

    grouped = df_year.groupby(year_col).agg({meta_col: 'mean', user_col: 'mean'}).reset_index().sort_values(year_col)

    # Plot 1:
    plt.figure(figsize=(10,5))
    plt.plot(grouped[year_col], grouped[meta_col], marker='o', label='Meta (média)')
    plt.plot(grouped[year_col], grouped[user_col], marker='o', label='User (média)')
    plt.xlabel('Ano')
    plt.ylabel('Score (média)')
    plt.title('Média Meta vs User por Ano')
    plt.legend()
    plt.grid(True)
    plt.show()

    # Plot 2
    fig, ax = plt.subplots(figsize=(10,5))
    x = grouped[year_col].astype(int)
    ax.bar(x.astype(str), grouped[meta_col], color='skyblue', label='Meta (média)')
    ax.set_xlabel('Ano')
    ax.set_ylabel('Meta (média)')

    ax2 = ax.twinx()
    ax2.plot(x.astype(str), grouped[user_col], color='red', marker='o', label='User (média)')
    ax2.set_ylabel('User (média)')

    ax.set_title('Climograma: Meta (barras) vs User (linha) por Ano')
    lines, labels = ax.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax.legend(lines+lines2, labels+labels2, loc='upper left')
    fig.tight_layout()
    plt.show()

    display(grouped.head(15))
```

```python
genres = df['Genre']   
votes = df ['Votes'] 
```

```python

genre_candidates = ['Genre']
votes_candidates = ['Votes']

genre_col = next((c for c in genre_candidates if c in df.columns), None)
votes_col = next((c for c in votes_candidates if c in df.columns), None)

if not genre_col or not votes_col:
    print('Não encontrei as colunas de gênero/votos. Colunas disponíveis:')
    print(df.columns.tolist())
else:
    df_gen = df[[genre_col, votes_col]].copy()

    df_gen['votes_clean'] = pd.to_numeric(df_gen[votes_col].astype(str).str.replace(r'[^0-9]', '', regex=True), errors='coerce')
    df_gen = df_gen.dropna(subset=['votes_clean'])

    df_gen['genre_list'] = df_gen[genre_col].astype(str).str.split('[,;/|]+')
    df_expl = df_gen.explode('genre_list')
    df_expl['genre_list'] = df_expl['genre_list'].str.strip()
    df_expl = df_expl[df_expl['genre_list'] != '']

    genre_votes = df_expl.groupby('genre_list')['votes_clean'].sum().reset_index()
    genre_votes = genre_votes.sort_values('votes_clean', ascending=False)

    top_n = 15
    top_genres = genre_votes.head(top_n)

    plt.figure(figsize=(10,6))
    sns.barplot(x='votes_clean', y='genre_list', data=top_genres, palette='viridis')
    plt.xlabel('Total de Votos')
    plt.ylabel('Gênero')
    plt.title(f'Total de Votos por Gênero (Top {top_n})')

    import matplotlib.ticker as mtick
    ax = plt.gca()
    ax.xaxis.set_major_formatter(mtick.FuncFormatter(lambda x, _: f'{int(x):,}'))

    plt.tight_layout()
    plt.show()

    display(top_genres)
```
